```python
#3 Alarme
import time
from datetime import datetime
import os

hora = input("Horário do remédio (HH:MM): ")

print(f"Alarme definido para {hora}...")

while True:
    agora = datetime.now().strftime("%H:%M")
    if agora == hora:
        print("\n⏰ HORA DE TOMAR O REMÉDIO! ⏰\n")

        # Faz barulho no terminal
        for i in range(5):
            print('\a')  # beep
            time.sleep(1)
        break

    time.sleep(1)
```

```python
#4 Menu hamburguer com opções
import os
import time

def limpar():
    os.system('cls' if os.name == 'nt' else 'clear')

def mostrar_menu_fechar():
    print("☰  (fechado)")
    print("Digite 'abrir' para mostrar as opções.")

def mostrar_menu_abrir():
    print("(aberto)")
    print("1 - Meu Perfil")
    print("2 - Minhas Receitas")
    print("3 - Lembretes")
    print("4 - Acompanhantes")
    print("5 - Histórico Médico")
    print("6 - Alergias e Condições")
    print("7 - Médicos e Clinicas")
    print("8 - Documwentos")
    print("9 - Configurações")
    print()

# Programa principal Menu Hamburger
while True:
    limpar()
    mostrar_menu_fechar()

    acao = input("\nComando: ").lower()

    if acao == "abrir":
        limpar()
        mostrar_menu_abrir()
        escolha = input("Escolha uma opção: ")

        if escolha == "1":
            print("Você entrou em: Meu Perfil")
        elif escolha == "2":
            print("Você entrou em: Minhas Receitas")
        elif escolha == "3":
            print("Você entrou em: Lembretes")
        elif escolha == "4":
            print("Acompanhantes")
        elif escolha == "5":
            print("Histórico Médico")
        elif escolha == "6":
            print("Alergias e Condições")
        elif escolha == "7":
            print("Médicos e Clinicas")
        elif escolha == "8":
            print("Documwentos")
        elif escolha == "9":
            print("Configurações")
            break
        else:
            print("Opção inválida!")

        input("\nPressione ENTER para continuar...")

    else:
        print("Comando desconhecido.")
        time.sleep(1.5)
```

```python
#5 Scroll lateral

import tkinter as tk
from tkinter import ttk

def criar_scroll_lateral():
    # Cria a janela principal
    root = tk.Tk()
    root.title("Scroll Lateral Simples")
    root.geometry("320x568") # Define o tamanho inicial da janela

    # 1. Cria um widget Canvas (tela) que será "rolado"
    canvas = tk.Canvas(root)

    # 2. Cria a barra de rolagem horizontal e especifica a orientação
    scrollbar_h = ttk.Scrollbar(root, orient=tk.HORIZONTAL, command=canvas.xview)

    # 3. Cria um Frame dentro do Canvas para conter os elementos que serão rolados
    # Este frame precisa ser grande o suficiente para que o scroll seja necessário
    scroll_frame = ttk.Frame(canvas)

    # Configura a função de rolagem para a barra de rolagem
    canvas.configure(xscrollcommand=scrollbar_h.set)

    # Adiciona elementos ao frame interno para demonstrar o scroll
    # Adicionamos um texto muito longo para forçar a rolagem lateral
    long_text = "Este é um exemplo de texto muito longo que se estende horizontalmente para demonstrar a funcionalidade de scroll lateral. Continuamos escrevendo para garantir que o texto ultrapasse a largura visível da janela."
    tk.Label(scroll_frame, text=long_text, wraplength=None).pack(padx=10, pady=10, side=tk.LEFT)

    # Empacota o canvas e a scrollbar na janela principal
    scrollbar_h.pack(side=tk.BOTTOM, fill=tk.X)
    canvas.pack(side=tk.TOP, fill=tk.BOTH, expand=True)

    # Adiciona o frame rolável dentro do canvas
    canvas.create_window((0, 0), window=scroll_frame, anchor="nw")

    # Configura o Canvas para redefinir a área de rolagem quando o frame interno muda de tamanho
    scroll_frame.bind(
        "<Configure>",
        lambda e: canvas.configure(
            scrollregion=canvas.bbox("all")
        )
    )

    # Inicia o loop principal da aplicação
    root.mainloop()

# Executa a função
if __name__ == "__main__":
    criar_scroll_lateral()
```
