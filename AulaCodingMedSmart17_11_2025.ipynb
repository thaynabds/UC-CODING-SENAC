```python
#3 Alarme
import time
from datetime import datetime
import os

hora = input("Horário do remédio (HH:MM): ")

print(f"Alarme definido para {hora}...")

while True:
    agora = datetime.now().strftime("%H:%M")
    if agora == hora:
        print("\n⏰ HORA DE TOMAR O REMÉDIO! ⏰\n")

        # Faz barulho no terminal
        for i in range(5):
            print('\a')  # beep
            time.sleep(1)
        break

    time.sleep(1)
```

```python
#4 Menu hamburguer com opções
import os
import time

def limpar():
    os.system('cls' if os.name == 'nt' else 'clear')

def mostrar_menu_fechar():
    print("☰  (fechado)")
    print("Digite 'abrir' para mostrar as opções.")

def mostrar_menu_abrir():
    print("(aberto)")
    print("1 - Meu Perfil")
    print("2 - Minhas Receitas")
    print("3 - Lembretes")
    print("4 - Acompanhantes")
    print("5 - Histórico Médico")
    print("6 - Alergias e Condições")
    print("7 - Médicos e Clinicas")
    print("8 - Documwentos")
    print("9 - Configurações")
    print()

# Programa principal Menu Hamburger
while True:
    limpar()
    mostrar_menu_fechar()

    acao = input("\nComando: ").lower()

    if acao == "abrir":
        limpar()
        mostrar_menu_abrir()
        escolha = input("Escolha uma opção: ")

        if escolha == "1":
            print("Você entrou em: Meu Perfil")
        elif escolha == "2":
            print("Você entrou em: Minhas Receitas")
        elif escolha == "3":
            print("Você entrou em: Lembretes")
        elif escolha == "4":
            print("Acompanhantes")
        elif escolha == "5":
            print("Histórico Médico")
        elif escolha == "6":
            print("Alergias e Condições")
        elif escolha == "7":
            print("Médicos e Clinicas")
        elif escolha == "8":
            print("Documwentos")
        elif escolha == "9":
            print("Configurações")
            break
        else:
            print("Opção inválida!")

        input("\nPressione ENTER para continuar...")

    else:
        print("Comando desconhecido.")
        time.sleep(1.5)
```
